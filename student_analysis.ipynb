# Student Performance Analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Load Data
The student performance dataset is loaded using pandas for analysis.

```python
df = pd.read_csv(r"C:\Users\suvar\student_data.csv", sep=";")
df.head()
```

## Dataset Description

The student performance dataset contains information about students such as gender, study time, and grades.  
This notebook analyzes the dataset to understand factors affecting student performance.

```python
df.shape
```

```python
df.columns
```

## Clean Data
In this step, the dataset is checked for missing values and duplicate records.

```python
df.isnull().sum()
```

```python
df = df.drop_duplicates()

print("Duplicates removed if any")
```

## Analyze Data
Basic analysis is performed to understand student performance, including average grade, study time relationship, and gender comparison.

```python
avg_grade = df["G3"].mean()

print("Average Final Grade:", avg_grade)
```

```python
high_score = df[df["G3"] > 15]

print("Number of students scoring above 15:", len(high_score))
```

```python
correlation = df["studytime"].corr(df["G3"])

print("Correlation between Study Time and Final Grade:", correlation)
```

```python
gender_performance = df.groupby("sex")["G3"].mean()

if gender_performance["M"] > gender_performance["F"]:
    print("Male students perform better on average.")
else:
    print("Female students perform better on average.")
```

## Visualize Data
Graphs are created to visualize grade distribution and relationships between variables.

```python
plt.hist(df["G3"], bins=10, edgecolor="black")

plt.title("Distribution of Final Grades")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.scatter(df["studytime"], df["G3"])

plt.title("Study Time vs Final Grade")
plt.xlabel("Study Time")
plt.ylabel("Final Grade (G3)")

plt.show()
```

```python
gender_performance = df.groupby("sex")["G3"].mean()

gender_performance.plot(kind="bar")

plt.title("Average Final Grade by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Final Grade")

plt.show()
```

## Conclusion

The dataset was analyzed using Python and pandas.

The average final grade of students was calculated.

Students who study more tend to score slightly better.

Gender comparison shows a small difference in average performance.

Visualizations helped understand the distribution and relationships in the dataset.
